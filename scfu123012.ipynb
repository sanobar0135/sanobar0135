```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("redSeaBream.png")

if img is None:
    print("Image not loaded")
else:
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img)
    plt.axis("off")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

plt.imshow(gray, cmap="gray")
plt.axis("off")
```

```python
import numpy as np

binary = np.zeros(gray.shape, dtype=np.uint8)
```

```python
threshold = 128
binary = np.where(gray > threshold, 255, 0).astype(np.uint8)

```

```python
import matplotlib.pyplot as plt

plt.imshow(binary, cmap="gray")
plt.axis("off")
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))

# Step 1: Dilation (fill gaps)
dilated = cv2.dilate(binary, kernel, iterations=1)

# Step 2: Erosion (remove noise)
clean_binary = cv2.erode(dilated, kernel, iterations=1)
```

```python
plt.imshow(clean_binary, cmap="gray")
plt.axis("off")

```

```python
cv2.imwrite("binary_mask.png", clean_binary)
```
