```python
from google.colab.patches import cv2_imshow
import cv2

# Read image in grayscale
img = cv2.imread("4 Sparrows.jpg", 0)

if img is None:
    print("Image not found ❌ Check filename.")
else:
    negative = 255 - img
    cv2_imshow(negative)
```

```python
_, thresh = cv2.threshold(img, 70, 255, cv2.THRESH_BINARY)
cv2_imshow(thresh)
```

```python
#gray level slicing without background
lower = 100
upper = 180

slice1 = np.zeros_like(img)

slice1[(img >= lower) & (img <= upper)] = 255
cv2_imshow(slice1)
```

```python
#gray level slicing with background
slice2 = img.copy()
slice2[(img >= lower) & (img <= upper)] = 255
cv2_imshow(slice2)
```

```python
#power law transform
gamma = 2
gamma_corrected = np.array(255 * (img/255)** gamma, dtype = 'uint8')
cv2_imshow(gamma_corrected)
```

```python
#log transform
#convert image to float to avoid overflow
img_float = img.astype(np.float32)

#compte constant
c = 255 / np.log(1 + np.max(img_float))

#apply log transform
log_transformed = c * np.log(1 + img_float)

#convert back to uint8
log_transformed = np.uint8(log_transformed)
cv2_imshow(log_transformed)
```

```python
#contrast stretching
min_val = np.min(img)
max_val = np.max(img)

contrast_stretch = (img - min_val) * (255 / (max_val - min_val))
contrast_stretch = np.array(contrast_stretch, dtype = 'uint8')
cv2_imshow(contrast_stretch)
```
